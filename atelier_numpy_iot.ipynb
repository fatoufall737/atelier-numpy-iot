Partie 1 : Listes Python vs tableaux NumPy
On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8
partie 1 ⁉
1) Créer une liste Python stockant ces températures;
2) Créer un tableau NumPy équivalent stockant ces températures ;
3) Effectuer une multiplication par 2 sur ces deux collections;
4) Quelle différence observe-t-on ?

```python
import numpy as np

np.random.seed(42)
```

```python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
temperatures_liste  #creation d'une liste contenant les temperatures
```

```python
#question 2 : crée le tableau NumPy équivalent, avec les mêmes valeurs :python

temperatures_array = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
temperatures_array
```

```python
#effectuer une multiplication par deux 2 sur ces deux collections
liste_x2 = temperatures_liste * 2
print("Liste * 2 :", liste_x2)

array_x2 = temperatures_array * 2
print("Array * 2 :", array_x2)
```

la difference observee est que :Sur une liste Python, l'opérateur * duplique la liste au lieu de multiplier les valeurs. Sur un tableau NumPy, * applique une multiplication élément par élément (vectorisée), ce qui est le comportement attendu pour du calcul scientifique.

Partie 7 – Concaténation
On suppose qu’on a un 2nd bâtiment avec 3 capteurs.  
1) Créer aléatoirement la matrice représentant les températures mesurées par les 3 capteurs
durant une journée
2) Déterminer de deux manières différentes la matrice qui fusionne les températures mesurées par
les 8 capteurs durant une journée

```python
#crée aléatoirement la matrice représentant les températures mesurées par les 3 capteurs du 2ᵉ bâtiment durant une journée.
np.random.seed(7)  # seed différent du bâtiment 1 pour avoir des données distinctes
temperatures_batiment2 = np.random.uniform(low=15, high=35, size=(3, 24))
temperatures_batiment2
```

```python
#ici on determine deux methodes differentes de déterminer la matrice qui fusionne les températures mesurées par les 8 capteurs (5 du bâtiment 1 + 3 du bâtiment 2) durant une journée.

# Méthode 1 : np.concatenate sur l'axe 0 (empilement des lignes/capteurs)
temperatures_8capteurs_v1 = np.concatenate((temperatures_5capteurs, temperatures_batiment2), axis=0)

# Méthode 2 : np.vstack, équivalent pour un empilement vertical
temperatures_8capteurs_v2 = np.vstack((temperatures_5capteurs, temperatures_batiment2))

print(temperatures_8capteurs_v1.shape, temperatures_8capteurs_v2.shape)
np.array_equal(temperatures_8capteurs_v1, temperatures_8capteurs_v2)
```

Partie 8 : Calcul scientifique
Considérons le tableau représentant les 24 températures du 1er bâtiment.
1) Pour les températures calculer le minimum, le maximum, la somme, la moyenne, la médiane, la
variance et l’écart-type
2) Standardiser le tableau des températures puis vérifier les valeurs de la moyenne et de l’écart
type du tableau standardisé
3) Normaliser (Min-Max) le tableau des températures puis vérifier les valeurs du tableau normalisé
4) Convertir le tableau initial en Fahrenheit

```python
#Pour les températures calculer le minimum, le maximum, la somme, la moyenne, la médiane, la variance et l’écart-type
# en utilisant les temperatures_jour (les 24 températures du 1ᵉʳ bâtiment).
print("Minimum   :", temperatures_jour.min())
print("Maximum   :", temperatures_jour.max())
print("Somme     :", temperatures_jour.sum())
print("Moyenne   :", temperatures_jour.mean())
print("Médiane   :", np.median(temperatures_jour))
print("Variance  :", temperatures_jour.var())
print("Écart-type:", temperatures_jour.std())
```

```python
# Standardisation du tableau, des températures puis on passe a la vérification des valeurs de la moyenne et de l’écart type du tableau standardisé
temperatures_standardisees = (temperatures_jour - temperatures_jour.mean()) / temperatures_jour.std()

print("Moyenne après standardisation   :", round(temperatures_standardisees.mean(), 10))
print("Écart-type après standardisation:", round(temperatures_standardisees.std(), 10))
# La moyenne doit être ~0 et l'écart-type ~1
```

```python
#Normalisation de (Min-Max) le tableau des températures puis vérifier les valeurs du tableau normalisé
temperatures_normalisees = (temperatures_jour - temperatures_jour.min()) / (temperatures_jour.max() - temperatures_jour.min())

print("Minimum après normalisation :", temperatures_normalisees.min())
print("Maximum après normalisation :", temperatures_normalisees.max())
# Toutes les valeurs doivent être comprises entre 0 et 1
```

```python
#Convertion du tableau initial en Fahrenheit
temperatures_fahrenheit = temperatures_jour * 9/5 + 32
temperatures_fahrenheit
```

La formule pour convertir Celsius → Fahrenheit est :

°F = °C × 9/5 + 32

C'est exactement ce que fait la ligne de code qu'on vient d'exécuter — elle applique cette formule à chacune des 24 températures du tableau d'un coup, grâce au calcul vectorisé de NumPy (pas besoin de boucle for).
